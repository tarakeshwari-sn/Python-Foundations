```python
import polars as pl
```

```python
# Show all columns when printing
pl.Config.set_tbl_cols(-1)
pl.Config.set_tbl_rows(100)
```

```python
# Read CSV
data = pl.read_csv("Airbnb_Open_Data.csv", infer_schema_length=1000)
```

```python
print(data.head(5))
print(data.schema)
```

```python
columns_to_drop=['reviews per month','review rate number','calculated host listings count','availability 365','house_rules','license','last review']
data=data.drop(columns_to_drop)
```

```python
data = data.unique(subset=['id'], keep="first")
```

```python
duplicate_count = data.select(pl.all().is_duplicated().sum())
print("Total duplicate rows:", duplicate_count)
```

```python
null_counts = data.select(pl.all().null_count())
print(null_counts)
```

```python
null_ratio = data.select([
    (pl.col(col).is_null().sum()/pl.len()).alias(col)
    for col in data.columns])
```

```python

print(null_ratio)
```

```python
cols_to_keep = [
    col for col in data.columns
    if data[col].null_count() / data.height < 0.5]
```

```python
data = data.select(cols_to_keep)
```

```python
if "instant_bookable" in data.columns:
    data=data.with_columns(pl.col("instant_bookable").cast(pl.Boolean).cast(pl.Int8))
```

```python

data.write_csv("Airbnb_Open_Data_Cleaned_Polars.csv")
print(data)
```
